```python
from typing import Dict, Optional


class Task:
    """Represents an individual project task with lifecycle tracking."""

    def __init__(self, task_id: int, description: str, assigned_to: str) -> None:
        self.task_id: int = task_id
        self.description: str = description
        self.status: str = "To Do"
        self.assigned_to: str = assigned_to

    def mark_in_progress(self) -> None:
        """Transitions task status to 'In Progress'."""
        self.status = "In Progress"

    def mark_complete(self) -> None:
        """Transitions task status to 'Complete'."""
        self.status = "Complete"

    def get_details(self) -> str:
        """Formats and returns formatted task metadata."""
        return (
            f"\n{'='*30}\n"
            f"ID:          {self.task_id}\n"
            f"Description: {self.description}\n"
            f"Status:      {self.status}\n"
            f"Assigned To: {self.assigned_to}\n"
            f"{'='*30}"
        )


class TaskManager:
    """Manages project tasks stored in a dictionary for O(1) lookups."""

    def __init__(self) -> None:
        self.project_tasks: Dict[int, Task] = {}

    def add_task(self, task_id: int, description: str, assigned_to: str) -> bool:
        """Adds a new task if the task_id does not already exist."""
        if task_id in self.project_tasks:
            return False
        self.project_tasks[task_id] = Task(task_id, description, assigned_to)
        return True

    def get_task(self, task_id: int) -> Optional[Task]:
        """Retrieves a task by ID, returning None if not found."""
        return self.project_tasks.get(task_id)


# --- Runnable Execution Block ---
if __name__ == "__main__":
    manager = TaskManager()

    # Add sample tasks
    manager.add_task(101, "Setup Database Schema", "Alice")
    manager.add_task(102, "Build API Endpoints", "Bob")

    # Update task state
    task_101 = manager.get_task(101)
    if task_101:
        task_101.mark_in_progress()

    # Output results to console
    print("=== TASK MANAGEMENT SYSTEM OUTPUT ===")
    for t_id in [101, 102]:
        task = manager.get_task(t_id)
        if task:
            print(task.get_details())
```
